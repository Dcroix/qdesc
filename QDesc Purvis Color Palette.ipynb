# QDesc PurVis (Purposeful Visuals) Color Palette Samples

```python
import qdesc as qd
import pandas as pd
```

```python
df = pd.DataFrame({
    'Category': ['Laptops', 'Desktops', 'Tablets', 'Smart_Phones', 'Television'],
    'Value': [23, 88, 55, 67, 34]
})
```

## Teal

```python
qd.purvis_col(df, 
           title = "Gadgets Sold for Current Year",
           category_col='Category',
           value_col='Value',
           xlabel= "Types of Gadgets", 
           ylabel = "Items Sold",
           mode = "top",
           top_n = 3, 
           palette = "teal")
```
