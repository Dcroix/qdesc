```python
import qdesc as qd
import pandas as pd
```

```python
df = pd.DataFrame({
    'Category': ['Laptops', 'Desktops', 'Tablets', 'Smart_Phones', 'Television'],
    'Value': [23, 88, 55, 67, 34]
})
```

## <font face = 'Trebuchet MS' color = '#1F2020'> Color Palette: <font/> <font face = 'Trebuchet MS' color = '#556890'> Teal <font/>

```python
qd.purvis_col(df, 
           title = "Gadgets Sold for Current Year",
           category_col='Category',
           value_col='Value',
           xlabel= "Types of Gadgets", 
           ylabel = "Items Sold",
           mode = "top",
           top_n = 3, 
           palette = "teal")
```
